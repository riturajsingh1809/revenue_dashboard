```python
# ==========================================
# REVENUE DASHBOARD PROJECT
# STARTING CODE - PART 1
# ==========================================

# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("="*60)
print("REVENUE DASHBOARD PROJECT")
print("="*60)

# ------------------------------------------
# Upload Dataset
# ------------------------------------------

from google.colab import files

uploaded = files.upload()

# ------------------------------------------
# Read Excel File
# ------------------------------------------

df = pd.read_excel("Sample_Dataset.xlsx")

print("\nDataset Loaded Successfully")

# ------------------------------------------
# Display Dataset
# ------------------------------------------

print("\nFirst Five Records")
print(df.head())

print("\nLast Five Records")
print(df.tail())

print("\nRandom Five Records")
print(df.sample(5))

# ------------------------------------------
# Shape
# ------------------------------------------

print("\nRows and Columns")
print(df.shape)

# ------------------------------------------
# Columns
# ------------------------------------------

print("\nColumn Names")
print(df.columns)

# ------------------------------------------
# Dataset Information
# ------------------------------------------

print("\nDataset Information")

df.info()

# ------------------------------------------
# Statistical Summary
# ------------------------------------------

print("\nStatistical Summary")

print(df.describe(include="all"))

# ------------------------------------------
# Missing Values
# ------------------------------------------

print("\nMissing Values")

print(df.isnull().sum())

# ------------------------------------------
# Duplicate Values
# ------------------------------------------

print("\nDuplicate Rows")

print(df.duplicated().sum())

# ------------------------------------------
# Remove Duplicate Rows
# ------------------------------------------

df = df.drop_duplicates()

print("\nDuplicate Removed Successfully")

# ------------------------------------------
# Convert Date Column
# ------------------------------------------

df["Date"] = pd.to_datetime(df["Date"])

print("\nDate Converted Successfully")

# ------------------------------------------
# Total Revenue
# ------------------------------------------

total_revenue = df["Revenue"].sum()

# ------------------------------------------
# Total Amount
# ------------------------------------------

total_amount = df["Amount"].sum()

# ------------------------------------------
# Average Revenue
# ------------------------------------------

average_revenue = df["Revenue"].mean()

# ------------------------------------------
# Maximum Revenue
# ------------------------------------------

maximum_revenue = df["Revenue"].max()

# ------------------------------------------
# Minimum Revenue
# ------------------------------------------

minimum_revenue = df["Revenue"].min()

# ------------------------------------------
# Total Transactions
# ------------------------------------------

total_transactions = len(df)

# ------------------------------------------
# Payment Status Analysis
# ------------------------------------------

payment_status = df["Payment_Status"].value_counts()

# ------------------------------------------
# Payment Method Analysis
# ------------------------------------------

payment_method = df["Payment_Method"].value_counts()

# ------------------------------------------
# Course Revenue
# ------------------------------------------

course_revenue = df.groupby("Course")["Revenue"].sum().sort_values(ascending=False)

# ------------------------------------------
# City Revenue
# ------------------------------------------

city_revenue = df.groupby("City")["Revenue"].sum().sort_values(ascending=False)

# ------------------------------------------
# KPI SUMMARY
# ------------------------------------------

print("\n")
print("="*60)
print("KPI SUMMARY")
print("="*60)

print("Total Revenue :", total_revenue)
print("Total Amount :", total_amount)
print("Average Revenue :", average_revenue)
print("Maximum Revenue :", maximum_revenue)
print("Minimum Revenue :", minimum_revenue)
print("Total Transactions :", total_transactions)

print("\nPayment Status")

print(payment_status)

print("\nPayment Method")

print(payment_method)

print("\nCourse Revenue")

print(course_revenue)

print("\nCity Revenue")

print(city_revenue)

print("\nProject Part 1 Completed Successfully")
```

```python
# ==========================================================
# REVENUE DASHBOARD PROJECT
# PART 2 : DATA ANALYSIS + VISUALIZATION
# ==========================================================

import matplotlib.pyplot as plt

print("="*60)
print("PART 2 : DATA ANALYSIS & VISUALIZATION")
print("="*60)

# ----------------------------------------------------------
# Revenue by Course
# ----------------------------------------------------------

course_revenue = df.groupby("Course")["Revenue"].sum().sort_values(ascending=False)

print("\nCourse Wise Revenue")
print(course_revenue)

plt.figure(figsize=(8,5))
course_revenue.plot(kind="bar")
plt.title("Course Wise Revenue")
plt.xlabel("Course")
plt.ylabel("Revenue")
plt.xticks(rotation=20)
plt.grid(True)
plt.show()

# ----------------------------------------------------------
# Revenue by City
# ----------------------------------------------------------

city_revenue = df.groupby("City")["Revenue"].sum().sort_values(ascending=False)

print("\nCity Wise Revenue")
print(city_revenue)

plt.figure(figsize=(8,5))
city_revenue.plot(kind="bar")
plt.title("City Wise Revenue")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.grid(True)
plt.show()

# ----------------------------------------------------------
# Payment Status
# ----------------------------------------------------------

payment_status = df["Payment_Status"].value_counts()

print("\nPayment Status")
print(payment_status)

plt.figure(figsize=(6,6))
payment_status.plot(kind="pie",autopct="%1.1f%%")
plt.ylabel("")
plt.title("Payment Status Distribution")
plt.show()

# ----------------------------------------------------------
# Payment Method
# ----------------------------------------------------------

payment_method = df["Payment_Method"].value_counts()

print("\nPayment Method")
print(payment_method)

plt.figure(figsize=(7,5))
payment_method.plot(kind="bar")
plt.title("Payment Method Analysis")
plt.xlabel("Payment Method")
plt.ylabel("Number of Payments")
plt.grid(True)
plt.show()

# ----------------------------------------------------------
# Daily Revenue Trend
# ----------------------------------------------------------

daily_revenue = df.groupby("Date")["Revenue"].sum()

print("\nDaily Revenue")
print(daily_revenue)

plt.figure(figsize=(10,5))
plt.plot(daily_revenue.index,daily_revenue.values,marker="o")
plt.title("Daily Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

# ----------------------------------------------------------
# Revenue Distribution
# ----------------------------------------------------------

plt.figure(figsize=(8,5))
plt.hist(df["Revenue"],bins=10)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

# ----------------------------------------------------------
# Amount Distribution
# ----------------------------------------------------------

plt.figure(figsize=(8,5))
plt.boxplot(df["Amount"])
plt.title("Amount Box Plot")
plt.ylabel("Amount")
plt.grid(True)
plt.show()

# ----------------------------------------------------------
# Top 10 Highest Revenue Records
# ----------------------------------------------------------

top10 = df.sort_values(by="Revenue",ascending=False)

print("\nTop 10 Revenue Records")
print(top10.head(10))

# ----------------------------------------------------------
# Success vs Failed Payments
# ----------------------------------------------------------

success = (df["Payment_Status"]=="Success").sum()
failed = (df["Payment_Status"]=="Failed").sum()
pending = (df["Payment_Status"]=="Pending").sum()

print("\nSummary")
print("Success :",success)
print("Failed :",failed)
print("Pending :",pending)

print("\nPART 2 COMPLETED SUCCESSFULLY")
print("="*60)
```

```python
# ==========================================================
# REVENUE DASHBOARD PROJECT
# PART 3 : KPI DASHBOARD + BUSINESS INSIGHTS
# ==========================================================

import matplotlib.pyplot as plt

print("="*70)
print("PART 3 : KPI DASHBOARD & BUSINESS INSIGHTS")
print("="*70)

# ==========================================================
# KPI CALCULATIONS
# ==========================================================

total_revenue = df["Revenue"].sum()
total_amount = df["Amount"].sum()
average_revenue = df["Revenue"].mean()
maximum_revenue = df["Revenue"].max()
minimum_revenue = df["Revenue"].min()
total_transactions = len(df)

success_payment = (df["Payment_Status"]=="Success").sum()
failed_payment = (df["Payment_Status"]=="Failed").sum()
pending_payment = (df["Payment_Status"]=="Pending").sum()

print("\n========== KPI SUMMARY ==========")
print("Total Revenue        :", total_revenue)
print("Total Amount         :", total_amount)
print("Average Revenue      :", round(average_revenue,2))
print("Maximum Revenue      :", maximum_revenue)
print("Minimum Revenue      :", minimum_revenue)
print("Total Transactions   :", total_transactions)
print("Success Payments     :", success_payment)
print("Failed Payments      :", failed_payment)
print("Pending Payments     :", pending_payment)

# ==========================================================
# COURSE WISE REVENUE
# ==========================================================

course = df.groupby("Course")["Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
course.plot(kind="bar")
plt.title("Course Wise Revenue Dashboard")
plt.xlabel("Course")
plt.ylabel("Revenue")
plt.grid(True)
plt.savefig("course_revenue_dashboard.png")
plt.show()

# ==========================================================
# CITY WISE REVENUE
# ==========================================================

city = df.groupby("City")["Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
city.plot(kind="bar")
plt.title("City Wise Revenue Dashboard")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.grid(True)
plt.savefig("city_revenue_dashboard.png")
plt.show()

# ==========================================================
# PAYMENT STATUS PIE CHART
# ==========================================================

status = df["Payment_Status"].value_counts()

plt.figure(figsize=(6,6))
status.plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.title("Payment Status")
plt.savefig("payment_status.png")
plt.show()

# ==========================================================
# DAILY REVENUE
# ==========================================================

daily = df.groupby("Date")["Revenue"].sum()

plt.figure(figsize=(12,5))
plt.plot(daily.index,daily.values,marker="o")
plt.title("Daily Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.grid(True)
plt.savefig("daily_revenue.png")
plt.show()

# ==========================================================
# TOP 10 CUSTOMERS
# ==========================================================

print("\n========== TOP 10 REVENUE RECORDS ==========")
print(df.sort_values("Revenue",ascending=False).head(10))

# ==========================================================
# BUSINESS INSIGHTS
# ==========================================================

print("\n========== BUSINESS INSIGHTS ==========")

print("1. Total Revenue Generated :", total_revenue)
print("2. Total Transactions :", total_transactions)
print("3. Highest Revenue :", maximum_revenue)
print("4. Lowest Revenue :", minimum_revenue)
print("5. Average Revenue :", round(average_revenue,2))
print("6. Most Popular Payment Method :", df["Payment_Method"].mode()[0])
print("7. Best Performing Course :", course.idxmax())
print("8. Best Revenue City :", city.idxmax())
print("9. Successful Payments :", success_payment)
print("10. Failed Payments :", failed_payment)

print("\nDashboard Images Saved Successfully")
print("course_revenue_dashboard.png")
print("city_revenue_dashboard.png")
print("payment_status.png")
print("daily_revenue.png")

print("\nPART 3 COMPLETED SUCCESSFULLY")
print("="*70)
```

```python
# ==========================================================
# REVENUE DASHBOARD PROJECT
# PART 4 : FINAL DASHBOARD + EXPORT
# ==========================================================

import pandas as pd
import matplotlib.pyplot as plt

print("="*70)
print("FINAL REVENUE DASHBOARD")
print("="*70)

# ==========================================================
# KPI SUMMARY TABLE
# ==========================================================

kpi = pd.DataFrame({
    "KPI":[
        "Total Revenue",
        "Total Amount",
        "Average Revenue",
        "Maximum Revenue",
        "Minimum Revenue",
        "Total Transactions",
        "Success Payments",
        "Failed Payments",
        "Pending Payments"
    ],
    "Value":[
        df["Revenue"].sum(),
        df["Amount"].sum(),
        round(df["Revenue"].mean(),2),
        df["Revenue"].max(),
        df["Revenue"].min(),
        len(df),
        (df["Payment_Status"]=="Success").sum(),
        (df["Payment_Status"]=="Failed").sum(),
        (df["Payment_Status"]=="Pending").sum()
    ]
})

print("\n========== KPI TABLE ==========\n")
print(kpi)

# ==========================================================
# EXPORT KPI TABLE
# ==========================================================

kpi.to_csv("KPI_Report.csv", index=False)

# ==========================================================
# EXPORT CLEAN DATASET
# ==========================================================

df.to_csv("Cleaned_Sample_Dataset.csv", index=False)

# ==========================================================
# COURSE SUMMARY
# ==========================================================

course_summary = df.groupby("Course").agg(
    Total_Revenue=("Revenue","sum"),
    Total_Amount=("Amount","sum"),
    Total_Students=("User_ID","count")
).sort_values(by="Total_Revenue",ascending=False)

print("\n========== COURSE SUMMARY ==========\n")
print(course_summary)

course_summary.to_csv("Course_Summary.csv")

# ==========================================================
# CITY SUMMARY
# ==========================================================

city_summary = df.groupby("City").agg(
    Total_Revenue=("Revenue","sum"),
    Total_Students=("User_ID","count")
).sort_values(by="Total_Revenue",ascending=False)

print("\n========== CITY SUMMARY ==========\n")
print(city_summary)

city_summary.to_csv("City_Summary.csv")

# ==========================================================
# PAYMENT SUMMARY
# ==========================================================

payment_summary = df.groupby("Payment_Method").agg(
    Revenue=("Revenue","sum"),
    Transactions=("User_ID","count")
)

print("\n========== PAYMENT SUMMARY ==========\n")
print(payment_summary)

payment_summary.to_csv("Payment_Summary.csv")

# ==========================================================
# TOP 10 RECORDS
# ==========================================================

top10 = df.sort_values(by="Revenue",ascending=False).head(10)

print("\n========== TOP 10 REVENUE RECORDS ==========\n")
print(top10)

top10.to_csv("Top10_Revenue.csv",index=False)

# ==========================================================
# FINAL INSIGHTS
# ==========================================================

print("\n========== FINAL BUSINESS INSIGHTS ==========")

print("✔️ Total Revenue :",df["Revenue"].sum())
print("✔️ Total Transactions :",len(df))
print("✔️ Average Revenue :",round(df["Revenue"].mean(),2))
print("✔️ Best Course :",course_summary.index[0])
print("✔️ Best City :",city_summary.index[0])
print("✔️ Best Payment Method :",df["Payment_Method"].mode()[0])
print("✔️ Highest Revenue :",df["Revenue"].max())
print("✔️ Lowest Revenue :",df["Revenue"].min())

print("\nFiles Created Successfully")

print("KPI_Report.csv")
print("Cleaned_Sample_Dataset.csv")
print("Course_Summary.csv")
print("City_Summary.csv")
print("Payment_Summary.csv")
print("Top10_Revenue.csv")

print("\nPART 4 COMPLETED SUCCESSFULLY")
print("="*70)
```
